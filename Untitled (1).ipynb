```python
# Restrict to adults 20+
df = df[df['RIDAGEYR'] >= 20].copy()
print(f"Adults 20+ N: {len(df):,}")
```

```python
# UACR (mg/g)
df['UACR'] = df['URXUMA'] / df['URXUCR'] * 100

# eGFR via CKD-EPI (sex-specific)
def calc_egfr(row):
    scr = row['LBXSCR']
    age = row['RIDAGEYR']
    sex = row['RIAGENDR']  # 1=Male, 2=Female
    if pd.isna(scr) or pd.isna(age):
        return np.nan
    if sex == 2:
        kappa, alpha = 0.7, -0.241
    else:
        kappa, alpha = 0.9, -0.302
    ratio = scr / kappa
    egfr = 142 * (min(ratio, 1)**alpha) * (max(ratio, 1)**-1.200) * (0.9938**age)
    if sex == 2:
        egfr *= 1.012
    return egfr

df['eGFR'] = df.apply(calc_egfr, axis=1)

print("Derived variables created.")
```

```python
def classify_ckm(row):
    # --- Stage 1 ---
    obesity     = row['BMXBMI'] >= 25.0 if pd.notna(row['BMXBMI']) else False
    prediabetes = (5.7 <= row['LBXGH'] < 6.5) if pd.notna(row['LBXGH']) else False
    fasting_pre = (100.0 <= row['LBXGLU'] < 126.0) if pd.notna(row['LBXGLU']) else False
    stage1 = obesity or prediabetes or fasting_pre

    # --- Stage 2 (each criterion independent) ---
    t2d = (row['DIQ010'] == 1 or
           (pd.notna(row['LBXGH'])  and row['LBXGH']  >= 6.5) or
           (pd.notna(row['LBXGLU']) and row['LBXGLU'] >= 126.0))

    htn = (row['BPQ020'] == 1 or
           (pd.notna(row['BPXOSY1']) and row['BPXOSY1'] >= 130.0) or
           (pd.notna(row['BPXODI1']) and row['BPXODI1'] >= 80.0))

    dyslipid = ((pd.notna(row['LBXTC'])  and row['LBXTC']  >= 200) or
                (pd.notna(row['LBDLDL']) and row['LBDLDL'] >= 130) or
                (pd.notna(row['LBDHDD']) and row['LBDHDD'] <  40)  or
                (pd.notna(row['LBXTR'])  and row['LBXTR']  >= 135))

    ckd = ((pd.notna(row['eGFR']) and row['eGFR'] < 60) or
           (pd.notna(row['UACR'])  and row['UACR'] >= 30))

    stage2 = t2d or htn or dyslipid or ckd

    # --- Stage 3 ---
    angina_proxy = row['CDQ001'] == 1 if pd.notna(row['CDQ001']) else False
    ckd_adv      = row['eGFR'] < 30   if pd.notna(row['eGFR'])   else False
    stage3 = angina_proxy or ckd_adv

    # --- Stage 4 ---
    hf     = row['MCQ160B'] == 1 if pd.notna(row['MCQ160B']) else False
    chd    = row['MCQ160C'] == 1 if pd.notna(row['MCQ160C']) else False
    mi     = row['MCQ160E'] == 1 if pd.notna(row['MCQ160E']) else False
    stroke = row['MCQ160F'] == 1 if pd.notna(row['MCQ160F']) else False
    pad    = row['MCQ160P'] == 1 if pd.notna(row['MCQ160P']) else False
    stage4 = hf or chd or mi or stroke or pad

    if stage4: return 4
    if stage3: return 3
    if stage2: return 2
    if stage1: return 1
    return 0


df['CKM_Stage'] = df.apply(classify_ckm, axis=1)
df['advanced_ckm'] = df['CKM_Stage'].isin([3, 4]).astype(int)

print(df['CKM_Stage'].value_counts().sort_index())
print("\nDistribution (%):")
print((df['CKM_Stage'].value_counts(normalize=True).sort_index() * 100).round(1))
print(f"\nAdvanced CKM (3+4): n={df['advanced_ckm'].sum()}, {df['advanced_ckm'].mean()*100:.1f}%")
```

```python
def classify_ckm(row):
    # --- Stage 1 ---
    obesity     = row['BMXBMI'] >= 25.0 if pd.notna(row['BMXBMI']) else False
    prediabetes = (5.7 <= row['LBXGH'] < 6.5) if pd.notna(row['LBXGH']) else False
    fasting_pre = (100.0 <= row['LBXGLU'] < 126.0) if pd.notna(row['LBXGLU']) else False
    stage1 = obesity or prediabetes or fasting_pre

    # --- Stage 2 (Soroush/AoU — each criterion independent) ---
    t2d     = (row['DIQ010'] == 1 or
               (pd.notna(row['LBXGH'])  and row['LBXGH']  >= 6.5) or
               (pd.notna(row['LBXGLU']) and row['LBXGLU'] >= 126.0))

    htn     = (row['BPQ020'] == 1 or
               (pd.notna(row['BPXOSY1']) and row['BPXOSY1'] >= 130.0) or
               (pd.notna(row['BPXODI1']) and row['BPXODI1'] >= 80.0))

    dyslipid = ((pd.notna(row['LBXTC'])  and row['LBXTC']  >= 200) or
                (pd.notna(row['LBDLDL']) and row['LBDLDL'] >= 130) or
                (pd.notna(row['LBDHDD']) and row['LBDHDD'] <  40)  or
                (pd.notna(row['LBXTR'])  and row['LBXTR']  >= 135))

    ckd     = ((pd.notna(row['eGFR']) and row['eGFR'] < 60) or
               (pd.notna(row['UACR'])  and row['UACR'] >= 30))

    stage2 = t2d or htn or dyslipid or ckd

    # --- Stage 3: proxy (CAC/ECHO/NTproBNP unavailable in NHANES) ---
    ckd_adv      = row['eGFR'] < 30   if pd.notna(row['eGFR'])   else False
    stage3 = ckd_adv

    # --- Stage 4: clinical CVD events ---
    hf     = row['MCQ160B'] == 1 if pd.notna(row['MCQ160B']) else False
    chd    = row['MCQ160C'] == 1 if pd.notna(row['MCQ160C']) else False
    mi     = row['MCQ160E'] == 1 if pd.notna(row['MCQ160E']) else False
    stroke = row['MCQ160F'] == 1 if pd.notna(row['MCQ160F']) else False
    pad    = row['MCQ160P'] == 1 if pd.notna(row['MCQ160P']) else False
    stage4 = hf or chd or mi or stroke or pad

    if stage4: return 4
    if stage3: return 3
    if stage2: return 2
    if stage1: return 1
    return 0

df['CKM_Stage'] = df.apply(classify_ckm, axis=1)
df['advanced_ckm'] = df['CKM_Stage'].isin([3, 4]).astype(int)

print(df['CKM_Stage'].value_counts().sort_index())
print("\nDistribution (%):")
print((df['CKM_Stage'].value_counts(normalize=True).sort_index() * 100).round(1))
print(f"\nAdvanced CKM (3+4): n={df['advanced_ckm'].sum()}, {df['advanced_ckm'].mean()*100:.1f}%")
```

```python
print("=== SENSITIVITY ANALYSIS ===\n")

# 1. Complete-case only
key_vars = ['BMXBMI','LBXGH','LBXGLU','eGFR','UACR',
            'BPXOSY1','BPXODI1','LBXTC','LBDLDL','LBDHDD','LBXTR']
df_complete = df.dropna(subset=key_vars).copy()
df_complete['CKM_Stage'] = df_complete.apply(classify_ckm, axis=1)
df_complete['advanced_ckm'] = df_complete['CKM_Stage'].isin([3,4]).astype(int)

print(f"Full sample N: {len(df):,}")
print(f"Complete-case N: {len(df_complete):,} ({len(df_complete)/len(df)*100:.1f}% retained)\n")
print("Complete-case weighted prevalence:")
for stage in range(5):
    mask = df_complete['CKM_Stage'] == stage
    pct = df_complete.loc[mask,'WTMECPRP'].sum() / df_complete['WTMECPRP'].sum() * 100
    print(f"  Stage {stage}: {pct:.1f}%")
adv = df_complete['advanced_ckm'] == 1
print(f"  Advanced CKM (3+4): {df_complete.loc[adv,'WTMECPRP'].sum()/df_complete['WTMECPRP'].sum()*100:.1f}%")

# 2. Strict CKD (both eGFR<60 AND UACR≥30)
def classify_ckm_strict(row):
    """Sensitivity: eGFR<60 AND UACR>=30 required for CKD at Stage 2.
       Stage 3 uses KDIGO very high risk (no angina, matching primary)."""
    obesity     = row['BMXBMI'] >= 25.0 if pd.notna(row['BMXBMI']) else False
    prediabetes = (5.7 <= row['LBXGH'] < 6.5) if pd.notna(row['LBXGH']) else False
    fasting_pre = (100.0 <= row['LBXGLU'] < 126.0) if pd.notna(row['LBXGLU']) else False
    stage1 = obesity or prediabetes or fasting_pre

    t2d = (row['DIQ010'] == 1 or
           (pd.notna(row['LBXGH'])  and row['LBXGH']  >= 6.5) or
           (pd.notna(row['LBXGLU']) and row['LBXGLU'] >= 126.0))
    htn = (row['BPQ020'] == 1 or
           (pd.notna(row['BPXOSY1']) and row['BPXOSY1'] >= 130.0) or
           (pd.notna(row['BPXODI1']) and row['BPXODI1'] >= 80.0))
    dyslipid = ((pd.notna(row['LBXTC'])  and row['LBXTC']  >= 200) or
                (pd.notna(row['LBDLDL']) and row['LBDLDL'] >= 130) or
                (pd.notna(row['LBDHDD']) and row['LBDHDD'] <  40)  or
                (pd.notna(row['LBXTR'])  and row['LBXTR']  >= 135))
    # Strict CKD: requires BOTH eGFR<60 AND UACR>=30
    ckd_strict = ((pd.notna(row['eGFR']) and row['eGFR'] < 60) and
                  (pd.notna(row['UACR'])  and row['UACR'] >= 30))
    stage2 = t2d or htn or dyslipid or ckd_strict

    # Stage 3: KDIGO very high risk (no angina)
    ckd_very_high = False
    if pd.notna(row['eGFR']):
        egfr = row['eGFR']
        uacr = row['UACR'] if pd.notna(row['UACR']) else None
        if egfr < 45:
            ckd_very_high = True
        elif 45 <= egfr < 60 and uacr is not None and uacr >= 30:
            ckd_very_high = True
        elif egfr >= 60 and uacr is not None and uacr > 300:
            ckd_very_high = True
    stage3 = ckd_very_high

    hf     = row['MCQ160B'] == 1 if pd.notna(row['MCQ160B']) else False
    chd    = row['MCQ160C'] == 1 if pd.notna(row['MCQ160C']) else False
    mi     = row['MCQ160E'] == 1 if pd.notna(row['MCQ160E']) else False
    stroke = row['MCQ160F'] == 1 if pd.notna(row['MCQ160F']) else False
    pad    = row['MCQ160P'] == 1 if pd.notna(row['MCQ160P']) else False
    stage4 = hf or chd or mi or stroke or pad

    if stage4: return 4
    if stage3: return 3
    if stage2: return 2
    if stage1: return 1
    return 0
  

df['CKM_Stage_Strict'] = df.apply(classify_ckm_strict, axis=1)
df['advanced_ckm_strict'] = df['CKM_Stage_Strict'].isin([3,4]).astype(int)

print("\n=== Strict CKD sensitivity (eGFR<60 AND UACR≥30) ===")
for stage in range(5):
    mask = df['CKM_Stage_Strict'] == stage
    pct = df.loc[mask,'WTMECPRP'].sum() / df['WTMECPRP'].sum() * 100
    print(f"  Stage {stage}: n={mask.sum():,}  weighted={pct:.1f}%")
adv_s = df['advanced_ckm_strict'] == 1
adv_m = df['advanced_ckm'] == 1
print(f"\nAdvanced CKM — main:   {df.loc[adv_m,'WTMECPRP'].sum()/df['WTMECPRP'].sum()*100:.1f}%")
print(f"Advanced CKM — strict: {df.loc[adv_s,'WTMECPRP'].sum()/df['WTMECPRP'].sum()*100:.1f}%")
```

```python
# --- IARC obesity-related cancer cohort ---
import numpy as np

cancer_cols = ['MCQ230A','MCQ230B','MCQ230C','MCQ230D']
IARC_SITES_NO_BREAST = {14, 15, 16, 17, 19, 23, 24, 32}
BREAST_CODE = 20
BREAST_AGE_CUTOFF = 55

def has_iarc_primary(row):
    age = row.get('RIDAGEYR', np.nan)
    for col in cancer_cols:
        val = row.get(col, np.nan)
        if pd.isna(val):
            continue
        code = int(val)
        if code in IARC_SITES_NO_BREAST:
            return True
        if code == BREAST_CODE and pd.notna(age) and age >= BREAST_AGE_CUTOFF:
            return True
    return False

def has_iarc_all_breast(row):
    all_sites = IARC_SITES_NO_BREAST | {BREAST_CODE}
    for col in cancer_cols:
        val = row.get(col, np.nan)
        if pd.notna(val) and int(val) in all_sites:
            return True
    return False

df['iarc_primary']    = df.apply(has_iarc_primary, axis=1)
df['iarc_all_breast'] = df.apply(has_iarc_all_breast, axis=1)

df_iarc      = df[df['iarc_primary']].copy()
df_iarc_sens = df[df['iarc_all_breast']].copy()

print(f"IARC primary cohort (breast age≥55): n={len(df_iarc)}")
print(f"IARC sensitivity (all breast):       n={len(df_iarc_sens)}")
```

```python
from scipy.stats import chi2_contingency
from itertools import combinations

# Post-hoc pairwise chi-square with Bonferroni correction

for groupvar, groupname in [('Sex', 'Sex'), ('Race', 'Race/Ethnicity')]:
    print(f"\n=== Post-hoc pairwise chi-square: {groupname} ===")
    groups = df[groupvar].dropna().unique()
    pairs = list(combinations(groups, 2))
    n_comparisons = len(pairs)
    alpha_corrected = 0.05 / n_comparisons
    print(f"Bonferroni-corrected α: 0.05 / {n_comparisons} = {alpha_corrected:.4f}\n")
    
    for g1, g2 in pairs:
        sub = df[df[groupvar].isin([g1, g2])]
        ct = pd.crosstab(sub[groupvar], sub['CKM_Stage'])
        chi2, p, dof, _ = chi2_contingency(ct)
        sig = "***" if p < alpha_corrected else "ns"
        print(f"  {g1} vs {g2}: χ²={chi2:.2f}, p={p:.4f} {sig}")
```

```python
# --- Multiple Imputation Sensitivity Analysis ---
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
import numpy as np

impute_vars = ['BMXBMI','LBXGH','LBXGLU','LBXTC','LBDLDL','LBDHDD','LBXTR','eGFR','UACR','CDQ001']

# Run 5 imputations, average results
n_imputations = 5
stage_results = []

for i in range(n_imputations):
    imp = IterativeImputer(random_state=i, max_iter=10)
    df_imp = df.copy()
    df_imp[impute_vars] = imp.fit_transform(df[impute_vars])
    
    df_imp['CKM_Stage_imp'] = df_imp.apply(classify_ckm, axis=1)
    df_imp['advanced_ckm_imp'] = df_imp['CKM_Stage_imp'].isin([3, 4]).astype(int)
    
    adv_wpct = df_imp.loc[df_imp['advanced_ckm_imp']==1, 'WTMECPRP'].sum() / df_imp['WTMECPRP'].sum() * 100
    stage_results.append(adv_wpct)
    print(f"  Imputation {i+1}: Advanced CKM = {adv_wpct:.1f}%")

print(f"\nMean across imputations: {np.mean(stage_results):.1f}%")
print(f"Range: {np.min(stage_results):.1f}% – {np.max(stage_results):.1f}%")
print(f"Complete-case estimate: 27.2%")
```
